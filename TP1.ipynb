# D-Generation de données de données et affichage
## D-1) génération des premières données
La fonction `make_blobs` crée un jeu de données artificiel : des points répartis en plusieurs groupes autour de centres choisis au hasard. Elle renvoie la matrice des données `X` (une ligne par point, une colonne par variable) et le vecteur `y`, qui contient le numéro du groupe de chaque point. On va générer 1000 (avec n_samples) données de 2 variables gaussiennes (n_features) en 4 groupes(centers).
### Que fait random_state ?
`random_state` fixe le point de départ (la graine) du générateur de nombres aléatoires. Quand on lui donne un nombre entier (0, 42 ou n'importe quel autre), on obtient les mêmes données à chaque exécution. Deux nombres différents donnent deux jeux de données différents, mais chacun reste identique d'une exécution à l'autre. Si on ne met pas `random_state`, les données changent à chaque exécution. Cela permet de reproduire exactement les mêmes résultats.

```python
from sklearn.datasets import make_blobs
X, y = make_blobs(n_samples = 1000, n_features = 2, centers = 4,random_state = 3)
```

### Visualisation des données générées
On représente les données sous forme de nuage de points avec `plt.scatter`. L'axe horizontal correspond à la variable 1 (première colonne de X) et l'axe vertical à la variable 2 (deuxième colonne de X). Chaque point est coloré selon son groupe (valeur de y), ce qui permet de voir les 4 groupes. Les axes sont fixés entre -15 et 15 avec `xlim` et `ylim`.

```python
plt.figure(figsize=(8,6))
plt.scatter(X[:,0],X[:,1],c=y)
plt.title("répartition des données de deux variables réparties en 4 groupes")
plt.xlim([-15,15])
plt.ylim([-15,15])
plt.xlabel("variables 1")
plt.ylabel("variables 2")
plt.show
```

## D-2) concaténation et affichage
### Je génère 100 données de deux variables en 2 groupes et 500 données de 2 variables en 3 groupes:

```python
X1, y1 = make_blobs(n_samples = 100, n_features = 2, centers=2, random_state = 42)
X2, y2 = make_blobs(n_samples = 500, n_features = 2, centers =3, random_state = 0)
```

### Je concatène les deux jeux de données et les numéros de classe pour fusionner les deux jeux de données
Je fusionne les deux jeux de données en un seul. J'utilise pour cela la bibliothèque `NumPy` (importée sous le nom np au tout début), qui sert à manipuler des tableaux de nombres. J'empile les deux matrices de données l'une sous l'autre avec `np.vstack`, et je met les deux vecteurs de numéros de classe bout à bout avec `np.hstack`.

```python
X_con=np.vstack((X1,X2))
Y_con=np.hstack((y1,y2))
```

### J'affiche les deux variables 

```python
plt.figure(figsize=(8,6))
plt.scatter(X_con[:,0],X_con[:,1],c=Y_con)
plt.title("répartition des données de deux variables réparties en 4 groupes")
plt.xlim([-15,15])
plt.ylim([-15,15])
plt.xlabel("variables 1")
plt.ylabel("variables 2")
plt.show
```
